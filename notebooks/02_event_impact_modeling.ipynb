# Event Impact Modeling

This notebook analyzes major financial inclusion events and their potential relationship with financial inclusion indicators.

Key events include:

- Telebirr Launch
- M-Pesa Ethiopia Launch
- Fayda Digital ID Rollout
- EthioPay Instant Payment System Launch
- NFIS-II Strategy Launch

These events are expected to influence:
- Account Ownership
- Mobile Money Adoption
- Digital Payments
- Financial Access

```python
events = df[df["record_type"] == "event"]

event_summary = (
    events.groupby("indicator")
    .size()
    .reset_index(name="Event Count")
)

event_summary
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

plt.bar(
    event_summary["indicator"],
    event_summary["Event Count"]
)

plt.xticks(rotation=45, ha="right")

plt.title("Major Financial Inclusion Events")

plt.tight_layout()

plt.show()
```

```python
timeline = pd.DataFrame({
    "Year":[2021,2022,2023,2024,2025],
    "Event":[
        "Telebirr Launch",
        "Safaricom Entry",
        "M-Pesa Launch",
        "Fayda Rollout",
        "NFIS-II"
    ]
})

timeline
```

```python
plt.figure(figsize=(10,4))

plt.plot(
    timeline["Year"],
    [1]*len(timeline),
    marker="o"
)

for i,row in timeline.iterrows():
    plt.text(
        row["Year"],
        1.02,
        row["Event"],
        rotation=45
    )

plt.yticks([])

plt.title(
    "Financial Inclusion Event Timeline"
)

plt.show()
```

```python
import pandas as pd

future_years = pd.DataFrame({
    "fiscal_year":[2025,2026,2027],
    "forecast":[54.241379,56.948276,59.655172]
})

future_years
```

```python
forecast_results = future_years.copy()

forecast_results["Conservative"] = (
    forecast_results["forecast"] * 0.90
)

forecast_results["Baseline"] = (
    forecast_results["forecast"]
)

forecast_results["Optimistic"] = (
    forecast_results["forecast"] * 1.10
)

forecast_results
```
